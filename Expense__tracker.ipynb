```python
# Smart Expense Tracker
# Made by Samiksha Verma - CSE Data Science
# Project 2

expenses = {
    "Food": 1200,
    "Travel": 800,
    "Books": 1500,
    "Internet": 500,
    "Shopping": 2000
}

print("--- Smart Expense Tracker ---")

total = sum(expenses.values())
print(f"Total Expenses: Rs. {total}")

highest = max(expenses, key=expenses.get)
print(f"Highest Expense: {highest} - Rs. {expenses[highest]}")

print("\n--- Expense Breakdown ---")
for category, amount in expenses.items():
    percent = (amount / total) * 100
    print(f"{category}: Rs. {amount} ({percent:.1f}%)")

# Savings check
budget = 7000
if total > budget:
    print(f"\nWarning: Over budget by Rs. {total - budget}!")
else:
    print(f"\nGood! You saved Rs. {budget - total}")

print(f"\nAverage Expense: Rs. {total / len(expenses):.2f}")
```
